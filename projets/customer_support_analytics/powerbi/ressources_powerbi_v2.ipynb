## 1.1 Comprendre les sources de données

> **🎯 À la fin de cette section**, tu sauras nommer chaque table, son **grain** (= ce que représente une ligne), et le rôle qu'elle joue dans le modèle.

### 📘 Concept clé — le « grain »

Le **grain** d'une table de faits, c'est ce que représente une ligne. Avant d'écrire la moindre formule, tu dois pouvoir compléter cette phrase :

> *« Une ligne de cette table = un(e) ____ »*

Si tu hésites, tes mesures seront fausses.

### Inventaire des 4 sources

| Fichier | Grain | Rôle | Volumétrie |
|---|---|---|---|
| `dim_agents.csv` | 1 agent (avec son tier, bureau, CSAT moyen historique) | Dimension | ~50 lignes |
| `dim_categories.csv` | 1 catégorie de ticket (avec son SLA strict ou non) | Dimension | ~12 lignes |
| `fact_tickets.csv` | 1 ticket (toute sa vie, du `created_at` à la résolution) | Fait principal | ~25 000 lignes |

### ⚠️ Piège fréquent — confondre `fact_tickets` et `tickets_risque_scores`

Le grain est identique (1 ligne = 1 ticket), mais les deux tables ne contiennent pas les mêmes tickets. `fact_tickets` contient **tout l'historique** (tickets résolus, fermés, rouverts). `tickets_risque_scores` ne contient que **les tickets sur lesquels le modèle ML s'est prononcé** — typiquement les ouverts du jour. Ne jamais joindre directement les deux, on les utilise séparément :

- KPIs SLA, CSAT, agents → **`fact_tickets`**
- Compteurs d'alertes ROUGE/ORANGE/JAUNE → **`tickets_risque_scores`**

# II — Modéliser les données

> **🎯 Objectif** : ton modèle a 5 tables (4 sources + Calendrier), 5 relations actives, et chaque dimension filtre les bonnes tables de faits.

Un modèle propre, c'est 80 % du dashboard fait. Les 48 mesures DAX qui suivront seront 3× plus simples si tu prends le temps ici.

## 2.1 Le schéma en étoile expliqué

### 📘 Concept clé — l'étoile vs le flocon

Un **schéma en étoile** place les tables de faits au centre et les dimensions tout autour. Les relations vont toujours **1 dimension → N faits**, jamais l'inverse, jamais entre deux dimensions.

*Pourquoi cette forme ?* Parce que Power BI optimise massivement ce pattern. Toute autre topologie (flocon, M2M, bidirectionnelle abusive) ralentit et complique le DAX.

### Diagramme du modèle AfriCare

```
                      +------------------+
                      |   Calendrier     |
                      +--------+---------+
                               | 1
                               | N
    +------------------+  +----+--------------------+
    |   dim_agents     |1 |                         |
    +---------+--------+--+    fact_tickets         |
              |              (table principale)     |
              | N         +--------+----------------+
    +---------+--------+ N|        | 1
    | dim_categories   |--+        |
    +------------------+1          | (pas de relation directe :
                                   |  les deux tables sont liees
                          +--------v-----------+   au Calendrier)
                          | tickets_risque_scores |
                          +-----------------------+
```

### 🚀 Pour aller plus loin

`tickets_risque_scores` n'est PAS reliée à `fact_tickets`. Les deux tables sont chacune reliées au `Calendrier` (via `created_at`/`date_created`) et le filtrage croisé passe par là. C'est volontaire : éviter une relation 1-1 entre deux faits qui crée des ambiguïtés.

## 2.2 Créer la table Calendrier

### 📘 Concept clé — pourquoi une table Calendrier ?

Toutes les fonctions de **time intelligence** (`SAMEPERIODLASTYEAR`, `PREVIOUSMONTH`, `DATEADD`...) exigent une table de dates :
1. **Continue** (pas de jours manquants),
2. **Marquée comme table de dates**,
3. **Reliée aux faits** par une relation 1→N.

Sans ces 3 conditions, tes comparaisons vs N-1 retourneront blank.

### 🔧 Création

**Modélisation** → **Nouvelle table** → coller :

```dax
Calendrier = 
VAR _start = DATE(2023,1,1)
VAR _end   = DATE(2025,12,31)
RETURN
ADDCOLUMNS(
    CALENDAR(_start, _end),
    "Annee",         YEAR([Date]),
    "Mois_Num",      MONTH([Date]),
    "Mois_Nom",      FORMAT([Date], "mmm", "fr-FR"),
    "Mois_Nom_Long", FORMAT([Date], "mmmm", "fr-FR"),
    "Annee_Mois",    FORMAT([Date], "yyyy-MM"),
    "Trimestre",     "T" & QUARTER([Date]),
    "Semaine",       WEEKNUM([Date], 2),
    "Jour_Semaine",  FORMAT([Date], "dddd", "fr-FR"),
    "Est_Weekend",   IF(WEEKDAY([Date], 2) >= 6, 1, 0)
)
```

### ✅ Vérification

Dans la vue **Données**, sélectionne `Calendrier` : tu dois voir 1 096 lignes (3 ans avec un bissextile en 2024) et 10 colonnes.

# IV — Construire les 48 mesures DAX

### 📘 Concept clé — qu'est-ce qu'une mesure DAX ?

**Analogie Excel.** Une mesure DAX, c'est comme une formule Excel **mais réévaluée à la volée pour chaque visuel** en fonction des filtres appliqués (slicer, ligne du tableau, segment du graphique). En Excel tu écris `=COUNT(B2:B100)` pour une plage figée. En DAX, `Nb Tickets Total = COUNTROWS(fact_tickets)` renverra :
- le total quand elle est dans une carte sans filtre,
- les tickets de Cocody quand elle est dans une cellule de tableau filtrée sur Cocody,
- les tickets de janvier quand elle est sur un axe temporel.

### Vue d'ensemble des 6 dossiers

| Dossier | Mesures | Rôle |
|---|---|---|
| `KPIs_GLOBAUX` | 16 | Compteurs SLA, CSAT, délais, risque |
| `SLA` | 6 | Breach strict, dépassement, évolution mois précédent |
| `Agents` | 9 | Performance, segmentation, couleurs Tier/Segment |
| `Alertes ML` | 7 | Alertes ROUGE/ORANGE/JAUNE, score risque |
| `Variations vs N-1` | 5 | Comparaison année précédente avec flèches Unicode |
| `Couleurs Variation` | 3 | Mise en forme conditionnelle |

### Comment créer une mesure et la classer

1. Clic droit sur `_Mesures` → **Nouvelle mesure**
2. Coller le code DAX
3. Bandeau **Outils de mesure** → champ **Dossier d'affichage** → taper le nom du dossier
4. Définir le **format** (nombre, pourcentage, monnaie)

## 4.1 Dossier `KPIs_GLOBAUX` (16 mesures)

> **🎯 Rôle du dossier** : poser le socle. Ces 16 mesures alimentent toute la page Vue d'ensemble.

### 📘 Avant le code — la logique métier

Toutes les mesures de comptage utilisent `COUNTROWS(fact_tickets)`, parfois enveloppé dans un `CALCULATE` qui filtre sur le `statut` ou un flag. Toutes les mesures de pourcentage utilisent `DIVIDE` (jamais `/`).

### Les 16 mesures

```dax
Nb Tickets Total = COUNTROWS(fact_tickets)
```

```dax
Nb Tickets Resolus = 
CALCULATE(
    COUNTROWS(fact_tickets),
    fact_tickets[statut] = "Resolu"
)
```

```dax
Taux Resolution % = DIVIDE([Nb Tickets Resolus], [Nb Tickets Total], 0)
```

```dax
Nb SLA Breach = 
CALCULATE(
    COUNTROWS(fact_tickets),
    fact_tickets[sla_breach] = 1
)
```

```dax
Taux SLA Breach % = DIVIDE([Nb SLA Breach], [Nb Tickets Total], 0)
```

```dax
Nb Backlog       = CALCULATE(COUNTROWS(fact_tickets), fact_tickets[in_backlog] = 1)
Taux Backlog %   = DIVIDE([Nb Backlog], [Nb Tickets Total], 0)
Nb Escalade      = CALCULATE(COUNTROWS(fact_tickets), fact_tickets[statut] = "Escalade")
Taux Escalade %  = DIVIDE([Nb Escalade], [Nb Tickets Total], 0)
```

```dax
Delai Premiere Reponse Moy (h) = AVERAGE(fact_tickets[first_response_heures])
Delai Resolution Moy (h)       = AVERAGE(fact_tickets[resolution_heures])
```

```dax
CSAT Moyen = 
AVERAGEX(
    FILTER(fact_tickets, NOT(ISBLANK(fact_tickets[csat]))),
    fact_tickets[csat]
)
```

```dax
Nb Tickets Rouverts  = CALCULATE(COUNTROWS(fact_tickets), fact_tickets[reopened] = 1)
Taux Reouverture %   = DIVIDE([Nb Tickets Rouverts], [Nb Tickets Total], 0)
Nb Tickets a Risque  = COUNTROWS(tickets_risque_scores)
Taux Risque %        = DIVIDE([Nb Tickets a Risque], [Nb Tickets Total], 0)
```

### ⚠️ Piège fréquent — `CSAT Moyen`

Beaucoup de tickets n'ont **pas de CSAT** (le client n'a pas répondu au questionnaire). Un simple `AVERAGE(fact_tickets[csat])` ignore les blanks par défaut et donne une moyenne correcte. Mais `SUM([csat]) / COUNT([id])` inclurait les blanks au dénominateur et dégraderait artificiellement la moyenne. Le pattern `AVERAGEX(FILTER(... NOT(ISBLANK(...))))` rend explicite la décision : on ne moyenne que les tickets avec CSAT renseigné.

### 🚀 Pour aller plus loin

`DIVIDE(num, den, 0)` avec le 3ᵉ argument à `0` force le retour à 0 quand le dénominateur est blank ou nul. Sans ce 3ᵉ argument, tu obtiendrais blank — utile parfois, gênant souvent.

## 4.2 Dossier `SLA` (6 mesures)

> **🎯 Rôle du dossier** : zoomer sur les engagements contractuels stricts. Sur AfriCare, certaines catégories ont un SLA *strict* (urgences médicales, fraudes) où tout dépassement coûte de l'argent.

```dax
Nb Breach SLA Strict = 
CALCULATE(
    COUNTROWS(fact_tickets),
    fact_tickets[sla_breach] = 1,
    dim_categories[sla_strict] = TRUE()
)
```

```dax
Taux Breach SLA Strict % = 
DIVIDE(
    [Nb Breach SLA Strict],
    CALCULATE(COUNTROWS(fact_tickets), dim_categories[sla_strict] = TRUE()),
    0
)
```

```dax
Depassement SLA Moy (h) = 
CALCULATE(
    AVERAGEX(fact_tickets, fact_tickets[resolution_heures] - fact_tickets[sla_heures]),
    fact_tickets[sla_breach] = 1
)
```

```dax
Ratio SLA Moyen = AVERAGEX(fact_tickets, DIVIDE(fact_tickets[resolution_heures], fact_tickets[sla_heures]))
```
*Format : `0.00"x"` — un ratio de `1.50x` veut dire qu'on a mis 50 % de plus que le SLA.*

```dax
Taux SLA Breach % PM = CALCULATE([Taux SLA Breach %], PREVIOUSMONTH(Calendrier[Date]))
```

```dax
Evolution Breach pp = ([Taux SLA Breach %] - [Taux SLA Breach % PM]) * 100
```
*Format : `+0.0;-0.0;0` — affiche par exemple `+2,3` ou `-1,5`.*

### 📘 Concept clé — pp (points de pourcentage) vs %

Si le `Taux SLA Breach %` passe de **40 %** à **42 %**, l'évolution est de **+2 pp** (points de pourcentage), pas **+5 %**. Confondre les deux est une faute fréquente qui décrédibilise un dashboard.

## 4.3 Dossier `Agents` (9 mesures)

> **🎯 Rôle du dossier** : segmenter les agents en **Top / Coach / Urgent** selon leur breach SLA et leur écart CSAT, puis générer les couleurs pour la mise en forme.

### 📘 Concept clé — la matrice 2×2 de l'agent

Deux dimensions :
- **Axe X** : `Taux SLA Breach %` (qualité de tenue des engagements)
- **Axe Y** : `Ecart CSAT Agent` (écart positif ou négatif vs le CSAT moyen global)

Trois segments :

| Conditions | Segment | Couleur fond | Couleur texte |
|---|---|---|---|
| Breach ≥ 55 % | **Urgent** (peu importe le CSAT) | `#FEE2E2` | `#EF4444` |
| Breach ≥ 40 % et Ecart CSAT < 0 | **Coach** | `#FEF3C7` | `#F59E0B` |
| Ecart CSAT ≥ 0 et Breach < 40 % | **Top** | `#DCFCE7` | `#10B981` |
| Sinon | **Coach** (par défaut) | `#FEF3C7` | `#F59E0B` |

```dax
Taux Breach Agent %    = [Taux SLA Breach %]
Taux Escalade Agent %  = [Taux Escalade %]
```

```dax
Ecart CSAT Agent = 
VAR _agentCsat  = [CSAT Moyen]
VAR _globalCsat = CALCULATE([CSAT Moyen], ALL(dim_agents))
RETURN _agentCsat - _globalCsat
```

```dax
Rang Agent SLA  = RANKX(ALL(dim_agents[agent_id]), CALCULATE([Taux SLA Breach %]), , ASC)
Rang Agent CSAT = RANKX(ALL(dim_agents[agent_id]), CALCULATE([CSAT Moyen]), , DESC)
```

```dax
Couleur Breach Agent = 
VAR _tx = [Taux SLA Breach %]
RETURN SWITCH(TRUE(),
    _tx >= 0.50, "#EF4444",
    _tx >= 0.40, "#F59E0B",
    "#10B981"
)
```

```dax
Segment Agent = 
VAR _ecart  = [Ecart CSAT Agent]
VAR _breach = [Taux SLA Breach %]
RETURN SWITCH(TRUE(),
    _breach >= 0.55, "Urgent",
    _breach >= 0.40 && _ecart < 0, "Coach",
    _ecart >= 0 && _breach < 0.40, "Top",
    "Coach"
)
```

```dax
Couleur Tier Agent = 
SWITCH(SELECTEDVALUE(dim_agents[tier]),
    "Tier 1", "#F3F4F6",
    "Tier 2", "#DCFCE7",
    "Tier 3", "#EDE9FE",
    "#F3F4F6"
)

Couleur Texte Tier Agent = 
SWITCH(SELECTEDVALUE(dim_agents[tier]),
    "Tier 1", "#6B7280",
    "Tier 2", "#10B981",
    "Tier 3", "#5B21B6",
    "#6B7280"
)
```

```dax
Couleur Segment Agent = 
SWITCH([Segment Agent],
    "Urgent", "#FEE2E2",
    "Coach",  "#FEF3C7",
    "Top",    "#DCFCE7",
    "#FEF3C7"
)

Couleur Texte Segment Agent = 
SWITCH([Segment Agent],
    "Urgent", "#EF4444",
    "Coach",  "#F59E0B",
    "Top",    "#10B981",
    "#F59E0B"
)
```

### ⚠️ Piège fréquent — le couple `Couleur fond` + `Couleur texte`

Pour les badges (chips) Tier et Segment, on utilise une **paire** : un fond pâle et un texte saturé. Si tu n'utilises qu'une couleur de fond, le texte sera illisible (texte noir par défaut sur fond ambre = OK ; mais texte noir sur fond rouge clair = sale). Toujours fournir les deux mesures et appliquer les deux mises en forme conditionnelles.

## 4.4 Dossier `Alertes ML` (7 mesures)

> **🎯 Rôle du dossier** : compter les alertes émises par le modèle ML, calculer un score moyen de risque, et afficher un titre dynamique.

### 📘 Concept clé — les 4 niveaux d'alerte

Le modèle ML attribue à chaque ticket ouvert un `score_risque` ∈ [0, 1] qui devient un `niveau_alerte` :
- **ROUGE** : score ≥ 0.80 — agir aujourd'hui
- **ORANGE** : 0.50 ≤ score < 0.80 — agir cette semaine
- **JAUNE** : 0.30 ≤ score < 0.50 — surveiller
- **VERT** : score < 0.30 — aucune action

Les compteurs ci-dessous utilisent `SEARCH("ROUGE", ...) > 0` plutôt que `= "ROUGE"` pour rester robuste si les libellés évoluent (ex: `ROUGE-CRITIQUE`).

```dax
Nb Alertes Total = COUNTROWS(tickets_risque_scores)
```

```dax
Nb Alertes ROUGE = 
CALCULATE(
    COUNTROWS(tickets_risque_scores),
    SEARCH("ROUGE", tickets_risque_scores[niveau_alerte], 1, 0) > 0
)

Nb Alertes ORANGE = 
CALCULATE(
    COUNTROWS(tickets_risque_scores),
    SEARCH("ORANGE", tickets_risque_scores[niveau_alerte], 1, 0) > 0
)

Nb Alertes JAUNE = 
CALCULATE(
    COUNTROWS(tickets_risque_scores),
    SEARCH("JAUNE", tickets_risque_scores[niveau_alerte], 1, 0) > 0
)
```

```dax
Score Risque Moy = 
CALCULATE(
    AVERAGEX(
        tickets_risque_scores,
        VALUE(SUBSTITUTE(tickets_risque_scores[score_risque], ".", ","))
    ),
    SEARCH("VERT", tickets_risque_scores[niveau_alerte], 1, 0) = 0
)
```

```dax
Titre Alertes = 
"Alertes ML - " & [Nb Alertes ROUGE] & " ROUGE - " 
    & [Nb Alertes ORANGE] & " ORANGE - " 
    & [Nb Alertes JAUNE] & " JAUNE"
```

```dax
Recall Modele = 0.796
```
*Format : `0.0%`. Le recall est figé en mesure simple — il vient du notebook ML après entraînement et passera à la valeur réelle quand le modèle sera réentraîné.*

### ⚠️ Piège fréquent — `score_risque` en String

Le CSV `tickets_risque_scores.csv` produit par scikit-learn écrit le score avec un point décimal (`0.78`). En locale FR, Power BI ne sait pas le convertir en `Double`. La parade : `VALUE(SUBSTITUTE([score_risque], ".", ","))` qui remplace le point par une virgule avant de convertir.

Alternative propre : reformater le CSV en amont avec un séparateur compatible. Si tu n'as pas la main sur le notebook ML, le `SUBSTITUTE` reste ta meilleure option.

## 4.5 Dossier `Variations vs N-1` (5 mesures)

> **🎯 Rôle du dossier** : pour chaque KPI principal, comparer à la même période l'année précédente avec une flèche ▲ ou ▼.

### 📘 Concept clé — la chaîne `SAMEPERIODLASTYEAR`

`SAMEPERIODLASTYEAR(Calendrier[Date])` reproduit **exactement la même fenêtre temporelle qu'aujourd'hui, mais 365 (ou 366) jours plus tôt**. Combinée à `CALCULATE`, elle te donne le KPI passé, et tu fais la différence.

Les flèches viennent de `UNICHAR(9650)` (▲) et `UNICHAR(9660)` (▼) — caractères Unicode universels.

```dax
Tickets vs N-1 = 
VAR _curr = [Nb Tickets Total]
VAR _prev = CALCULATE([Nb Tickets Total], SAMEPERIODLASTYEAR(Calendrier[Date]))
VAR _pct  = DIVIDE(_curr - _prev, _prev)
VAR _fmt  = FORMAT(_pct, "+0.0%;-0.0%")
RETURN SWITCH(TRUE(),
    _pct > 0, UNICHAR(9650) & " " & _fmt & " vs N-1",
    _pct < 0, UNICHAR(9660) & " " & _fmt & " vs N-1",
    UNICHAR(9650) & " 0,0% vs N-1"
)
```

```dax
SLA Breach vs N-1 = 
VAR _curr  = [Taux SLA Breach %]
VAR _prev  = CALCULATE([Taux SLA Breach %], SAMEPERIODLASTYEAR(Calendrier[Date]))
VAR _delta = (_curr - _prev) * 100
VAR _fmt   = FORMAT(_delta, "+0.0;-0.0") & "pp"
RETURN SWITCH(TRUE(),
    _delta > 0, UNICHAR(9650) & " " & _fmt & " vs N-1",
    _delta < 0, UNICHAR(9660) & " " & _fmt & " vs N-1",
    UNICHAR(9650) & " 0,0pp vs N-1"
)
```

```dax
Delai vs N-1 = 
VAR _curr  = [Delai Premiere Reponse Moy (h)]
VAR _prev  = CALCULATE([Delai Premiere Reponse Moy (h)], SAMEPERIODLASTYEAR(Calendrier[Date]))
VAR _delta = _curr - _prev
VAR _fmt   = FORMAT(_delta, "+0.0;-0.0") & "h"
RETURN SWITCH(TRUE(),
    _delta > 0, UNICHAR(9650) & " " & _fmt & " vs N-1",
    _delta < 0, UNICHAR(9660) & " " & _fmt & " vs N-1",
    UNICHAR(9650) & " 0,0h vs N-1"
)
```

```dax
Nb Categories >50% Breach = 
COUNTROWS(
    FILTER(VALUES(dim_categories[nom]), [Taux SLA Breach %] > 0.50)
)
```

```dax
Alerte Contractuelle = 
VAR _nb = [Nb Categories >50% Breach]
RETURN IF(
    _nb > 0,
    "Billing SLA strict : " & _nb & " categories depassent 50% de breach. Risque de penalites contractuelles.",
    BLANK()
)
```

### ⚠️ Piège fréquent — sémantique inversée pour SLA Breach

Pour `Tickets vs N-1`, une hausse est **bonne** (plus d'activité, signe d'usage). Pour `SLA Breach vs N-1` et `Delai vs N-1`, une hausse est **mauvaise** (on dégrade). C'est pourquoi les couleurs (dossier suivant) sont inversées.

# V — Créer les colonnes calculées

> **🎯 Objectif** : ajouter les flags binaires utilisés par le modèle ML et la colonne date pure pour relier les sessions au Calendrier.

### 📘 Concept clé — colonne calculée vs mesure

| | Colonne calculée | Mesure |
|---|---|---|
| **Évaluée** | Au refresh, ligne par ligne | À chaque rendu de visuel |
| **Stockée** | Dans la table | Nulle part |
| **Filtre / légende ?** | Oui | Non |
| **Coût mémoire** | + | 0 |
| **Coût CPU au visuel** | 0 | + |

**Règle de pouce** — si tu veux la valeur en **slicer, légende, axe**, c'est une **colonne calculée**. Sinon c'est une **mesure**.

## 5.1 Les flags risque (binaires sur `fact_tickets`)

> **🎯 À la fin de cette section**, tu auras 6 colonnes binaires (0/1) qui servent au modèle ML et à des filtres rapides.

**Vue Données → table `fact_tickets` → Nouvelle colonne** pour chacune :

```dax
ticket_at_risk = IF(fact_tickets[ratio_sla] >= 0.8 && fact_tickets[statut] <> "Resolu", 1, 0)
```
*Un ticket non résolu qui a déjà consommé 80 % de son SLA = à risque.*

```dax
is_high_priority  = IF(fact_tickets[priorite] >= 3, 1, 0)
is_long_wait      = IF(fact_tickets[first_response_heures] > 4, 1, 0)
is_reopened       = IF(fact_tickets[reopened] = 1, 1, 0)
est_weekend       = IF(WEEKDAY(fact_tickets[created_at], 2) >= 6, 1, 0)
heure_hors_bureau = IF(HOUR(fact_tickets[created_at]) < 8 || HOUR(fact_tickets[created_at]) >= 18, 1, 0)
```

### 📘 Pourquoi des `1/0` plutôt que des `TRUE/FALSE` ?

Power BI gère mal les booléens dans les agrégats : `MAX(TRUE/FALSE)` ne renvoie pas ce qu'on attend. Avec des entiers 0/1, `SUM` te donne le compte direct, `AVERAGE` te donne le pourcentage. C'est aussi le format attendu par scikit-learn dans le notebook ML.

## 5.2 La colonne date pure `date_created`

> **🎯 À la fin de cette section**, le slicer Année / Mois sur la page Vue d'ensemble filtrera correctement les tickets.

### 📘 Le problème

`fact_tickets[created_at]` est un `DateTime` à la milliseconde. Or, pour relier au Calendrier, il faut une `Date` pure. Sans cette colonne, la relation Calendrier ↔ fact_tickets est impossible et tous les filtres temporels échouent.

### Le DAX

**Vue Données → table `fact_tickets` → Nouvelle colonne** :

```dax
date_created = 
DATE(
    YEAR(fact_tickets[created_at]),
    MONTH(fact_tickets[created_at]),
    DAY(fact_tickets[created_at])
)
```

Puis crée la relation `Calendrier[Date]` (1) → `fact_tickets[date_created]` (N).

### 🚀 Pour aller plus loin

Pourquoi ne pas faire `DATEVALUE([created_at])` qui ferait pareil en plus court ? Parce que `DATEVALUE` rate certains formats locaux (FR vs US) et plante silencieusement. Le pattern `DATE(YEAR, MONTH, DAY)` est universel.

### 📘 Le `Recall Modele` à 79.6 %

Le recall mesure la proportion de tickets *réellement* à risque que le modèle a su détecter. À 79.6 %, sur 100 tickets qui dépasseront leur SLA, le modèle en flag 80. Les 20 manqués sont la « zone aveugle ». L'afficher honnêtement sur le dashboard maintient la confiance des opérateurs.

## 9.2 Pièges fréquents et solutions

| Symptôme | Cause | Correction |
|---|---|---|
| `SAMEPERIODLASTYEAR` renvoie blank | `Calendrier` non marquée comme table de dates | Outils de table → Marquer comme table de dates |
| `Score Risque Moy` plante | `score_risque` est en String avec point décimal | `VALUE(SUBSTITUTE([score_risque], ".", ","))` |
| Le scatter refuse `Segment Agent` en légende | Une mesure ne peut pas être une légende | Convertir en **colonne calculée** sur `dim_agents` |
| `SELECTEDVALUE` renvoie blank dans un total | Plusieurs valeurs dans le contexte | Remplacer par `MAX` ou conserver le blank côté visuel |
| `DIVIDE` renvoie blank au lieu de 0 | Voulu par défaut | `DIVIDE(num, den, 0)` pour forcer le 0 |
| `Evolution Breach pp` reste à 0 | Filtre Mois absent ou table de dates non marquée | Vérifier la slicer et la marque table de dates |
| Les flèches ▲ ▼ sont en carré | Police par défaut sans Unicode | Forcer **Segoe UI Symbol** sur le champ |
| Le tableau agents n'a pas la couleur de fond | Mise en forme conditionnelle non appliquée | Format → Cellules → fx → Mettre en forme par valeur du champ → choisir `Couleur Segment Agent` |

## 9.4 Annexes

### Règles DAX universelles à graver

1. Toujours `DIVIDE` (jamais `/`)
2. Pattern `VAR ... RETURN` même pour les mesures simples
3. Utiliser `SELECTEDVALUE` ou `MAX` au lieu de `VALUES` quand on attend une valeur unique
4. Dans `CALCULATE`, écrire `Table[Col] = "value"` plutôt que `Mesure = "value"`
5. Pour les variations en pp, faire une soustraction, pas un `DIVIDE`
6. `SEARCH(.., col, 1, 0) > 0` plutôt que `col = "..."` quand les libellés peuvent évoluer
