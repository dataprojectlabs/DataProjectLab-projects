## 2.1 Le schéma en étoile expliqué

> **🎯 À la fin de cette section**, tu sauras dessiner un schéma en étoile et expliquer pourquoi il bat tous les autres modèles en BI.

### 📘 Concept clé — l'étoile vs le flocon

Un **schéma en étoile** place les tables de faits au centre et les dimensions tout autour. Les relations sont toujours **1 dimension → N faits**, jamais l'inverse, jamais entre deux dimensions.

*Pourquoi cette forme ?* Parce que Power BI optimise massivement ce pattern. Toute autre topologie (flocon, M2M, relations bidirectionnelles abusives) ralentit le rapport et complique le DAX.

### Diagramme du modèle ShopAfrica+

Le modèle compte **8 tables** : 1 calendrier, 2 dimensions, 4 faits, 1 table de segmentation client.

```
                          +------------------+
                          |   Calendrier     |
                          |   (table date)   |
                          +--------+---------+
                                   | 1
                                   |
         +-----------------+       | N        +-----------------+
         |  dim_customers  | 1---N |     N---1|   dim_products  |
         +--------+--------+       |          +--------+--------+
                  |                |                   |
                  | N              |                 N |
             +----+----------------+-------------------+------+
             |                                                |
             |              fact_orders (commande)            |
             |                                                |
             +-----------------------+------------------------+
                                     | 1
                                     |
                                   N |
                            +--------+--------+
                            | fact_order_items|
                            +-----------------+

         +-----------------+
         |  fact_reviews   | (lié à dim_customers via customer_id,
         +-----------------+  à dim_products via product_id,
                              et au Calendrier via date_review)

         +-----------------+
         |  fact_web_logs  | (lié au Calendrier via session_date,
         +-----------------+  et à dim_customers via customer_id)

         +------------------------+
         |  clients_rfm_segments  | (lié à dim_customers via customer_id)
         +------------------------+
```

Quand un visuel est lent ou qu'une mesure ne renvoie pas ce que tu attends, retourne **toujours** au modèle avant de blâmer le DAX. 9 fois sur 10, le bug est une relation manquante ou une cardinalité mal réglée.

## 2.2 Créer la table Calendrier

> **🎯 À la fin de cette section**, tu disposes d'une table Calendrier maison, plus rapide et plus flexible que celle générée automatiquement.

### 📘 Concept clé — pourquoi une table Calendrier ?

Toutes les fonctions de **time intelligence** (`SAMEPERIODLASTYEAR`, `DATEADD`, `DATESYTD`...) exigent une table de dates :
1. **Continue** (pas de jours manquants),
2. **Marquée comme table de dates**,
3. **Reliée aux faits** par une relation 1→N.

Si l'une de ces 3 conditions manque, tes comparaisons vs N-1 retourneront blank.

### 🔧 Création

**Modélisation** → **Nouvelle table** → coller le code suivant :

```dax
Calendrier = 
VAR _start = DATE(2023,1,1)
VAR _end   = DATE(2025,12,31)
RETURN
ADDCOLUMNS(
    CALENDAR(_start, _end),
    "Annee",        YEAR([Date]),
    "Mois_Num",     MONTH([Date]),
    "Mois_Nom",     FORMAT([Date], "mmmm", "fr-FR"),
    "Trimestre",    "T" & QUARTER([Date]),
    "AnneeMois",    FORMAT([Date], "yyyy-MM"),
    "Jour_Num",     WEEKDAY([Date], 2),
    "Jour_Nom",     FORMAT([Date], "dddd", "fr-FR")
)
```

### ✅ Vérification

L'icône de la table `Calendrier` change : un petit calendrier apparaît en surimpression de l'icône de table. Sans cette étape, `SAMEPERIODLASTYEAR(Calendrier[Date])` ne renverra rien.

### 📘 Concept clé — pourquoi cette case à cocher est si importante

Power BI a besoin de savoir quelle colonne représente la *vraie* date de référence pour le moteur de time intelligence. Sans cette désignation, il fait des suppositions et certaines fonctions cessent de marcher silencieusement (sans message d'erreur). C'est le bug n°1 des débutants en DAX.

### 📘 Concept clé — qu'est-ce qu'une mesure DAX ?

**Analogie Excel.** Une mesure DAX, c'est comme une formule Excel **mais réévaluée à la volée pour chaque visuel** en fonction des filtres appliqués (slicer, ligne du tableau, segment du graphique). En Excel tu écris `=SUM(B2:B100)` pour une plage figée. En DAX, tu écris `Total CA = SUM(fact_order_items[line_revenue])` et la même formule renverra :
- le CA total quand elle est posée dans une carte sans filtre,
- le CA d'Abidjan quand elle est dans une cellule de tableau filtrée sur Abidjan,
- le CA mensuel quand elle est sur un axe temporel.

## 4.1 Dossier `01 — KPIs de base` (9 mesures)

> **🎯 Rôle du dossier** : poser le socle. Ces 9 mesures alimenteront 80 % du dashboard.

### 📘 Avant le code — la logique métier

Toutes ces mesures suivent la même règle : **on ne compte que les commandes livrées**. Une commande annulée ou retournée fausse le CA. C'est la décision métier prise avec le client ShopAfrica+.

```dax
Total CA = 
CALCULATE(
    SUM(fact_order_items[line_revenue]),
    fact_orders[order_status] = "Livree"
)
```

```dax
Total Commandes = 
CALCULATE(
    DISTINCTCOUNT(fact_orders[order_id]),
    fact_orders[order_status] = "Livree"
)
```

```dax
Total Clients = DISTINCTCOUNT(fact_orders[customer_id])
```

```dax
Total Produits Vendus = 
CALCULATE(
    SUM(fact_order_items[quantity]),
    fact_orders[order_status] = "Livree"
)
```

```dax
Panier Moyen = 
DIVIDE([Total CA], [Total Commandes])
```

```dax
Marge Totale = 
CALCULATE(
    SUMX(
        fact_order_items,
        fact_order_items[line_revenue] - (fact_order_items[quantity] * RELATED(dim_products[cout_unitaire]))
    ),
    fact_orders[order_status] = "Livree"
)
```

```dax
Taux Marge = DIVIDE([Marge Totale], [Total CA])
```

```dax
Frais Livraison Total = 
CALCULATE(
    SUM(fact_orders[shipping_fee]),
    fact_orders[order_status] = "Livree"
)
```

```dax
Note Moyenne = AVERAGE(fact_order_items[product_rating])
```

### ⚠️ Piège fréquent — `DIVIDE` vs `/`

Pourquoi `DIVIDE([Marge], [CA])` plutôt que `[Marge] / [CA]` ? Parce que la division par zéro renverra `Infinity` avec `/` (et plantera ton visuel) alors que `DIVIDE` renvoie un blank propre. **Règle absolue : toute division en DAX se fait avec `DIVIDE`.**

### 🚀 Pour aller plus loin

`SUMX` itère ligne à ligne (lent sur des millions de lignes). Si tu trouves que `Marge Totale` est lente, ajoute une colonne calculée `[line_margin]` dans `fact_order_items` et fais un simple `SUM` dessus.

Le pattern `VAR _x = ... RETURN ...` a deux avantages :
1. **Performance** : la variable est calculée une seule fois, même si tu la réutilises 10 fois.

Adopte ce pattern par défaut, même pour des mesures simples avec une seule étape : tes corrections futures te remercieront.

## 4.3 Dossier `03 — Variations vs N-1` (17 mesures)

> **🎯 Rôle du dossier** : pour chaque KPI principal, calculer la même valeur à la même période l'année précédente, puis la variation en pourcentage.

### 📘 Concept clé — `SAMEPERIODLASTYEAR` en une phrase

`SAMEPERIODLASTYEAR(Calendrier[Date])` reproduit **exactement la même fenêtre temporelle qu'aujourd'hui, mais 365 (ou 366) jours plus tôt**. Si je suis sur janvier-mars 2025, elle me renvoie janvier-mars 2024. Si je suis sur le 7 mai 2025, elle me renvoie le 7 mai 2024.

### Pattern universel

```dax
[KPI] N-1 = CALCULATE([KPI], SAMEPERIODLASTYEAR(Calendrier[Date]))
[Variation KPI] = DIVIDE([KPI] - [KPI] N-1, [KPI] N-1)
```

### Les 17 mesures

```dax
CA N-1            = CALCULATE([Total CA],            SAMEPERIODLASTYEAR(Calendrier[Date]))
Commandes N-1     = CALCULATE([Total Commandes],     SAMEPERIODLASTYEAR(Calendrier[Date]))
Clients N-1       = CALCULATE([Total Clients],       SAMEPERIODLASTYEAR(Calendrier[Date]))
Panier Moyen N-1  = CALCULATE([Panier Moyen],        SAMEPERIODLASTYEAR(Calendrier[Date]))
Marge N-1         = CALCULATE([Marge Totale],        SAMEPERIODLASTYEAR(Calendrier[Date]))
Taux Marge N-1    = CALCULATE([Taux Marge],          SAMEPERIODLASTYEAR(Calendrier[Date]))
Note Moyenne N-1  = CALCULATE([Note Moyenne],        SAMEPERIODLASTYEAR(Calendrier[Date]))
Taux Retour N-1   = CALCULATE([Taux Retour],         SAMEPERIODLASTYEAR(Calendrier[Date]))
```

Puis les 9 variations :

```dax
Variation CA           = DIVIDE([Total CA] - [CA N-1], [CA N-1])
Variation Commandes    = DIVIDE([Total Commandes] - [Commandes N-1], [Commandes N-1])
Variation Clients      = DIVIDE([Total Clients] - [Clients N-1], [Clients N-1])
Variation Panier       = DIVIDE([Panier Moyen] - [Panier Moyen N-1], [Panier Moyen N-1])
Variation Marge        = DIVIDE([Marge Totale] - [Marge N-1], [Marge N-1])
Variation Taux Marge   = [Taux Marge] - [Taux Marge N-1]
Variation Note         = [Note Moyenne] - [Note Moyenne N-1]
Variation Taux Retour  = [Taux Retour] - [Taux Retour N-1]
Variation Frais        = DIVIDE([Frais Livraison Total] - CALCULATE([Frais Livraison Total], SAMEPERIODLASTYEAR(Calendrier[Date])), CALCULATE([Frais Livraison Total], SAMEPERIODLASTYEAR(Calendrier[Date])))
```

### ⚠️ Piège fréquent — variations relatives vs absolues

Pour `Variation Taux Marge`, on fait une **soustraction de pourcentages** (différence en points), pas un `DIVIDE`. Sinon une marge passant de 20 % à 22 % afficherait +10 % au lieu de +2 pp. Cette nuance est ce qui sépare un dashboard amateur d'un dashboard pro.

## 4.4 Dossier `04 — Couleurs Variation` (18 mesures)

> **🎯 Rôle du dossier** : alimenter les mises en forme conditionnelles. Chaque variation a sa mesure de couleur dédiée.

### 📘 Concept clé — sémantique inversée

Pour la plupart des KPIs : **↑ = vert (bon), ↓ = rouge (mauvais)**.  
Mais pour `Taux Retour`, `Taux Annulation`, `Frais Livraison`, `Délai Livraison` : **↑ = rouge (mauvais), ↓ = vert (bon)**.

Tu créeras donc deux familles de mesures couleur, avec la même structure mais les seuils inversés.

### Pattern famille A — sens normal (CA, marge, etc.)

```dax
Couleur Variation CA = 
SWITCH(TRUE(),
    [Variation CA] > 0.02,  "#1D9E75",
    [Variation CA] < -0.02, "#E24B4A",
    "#888780"
)
```

*(Vert au-dessus de +2 %, rouge en-dessous de -2 %, gris dans la zone neutre.)*

### Pattern famille B — sens inversé (retour, annulation)

```dax
Couleur Variation Taux Retour = 
SWITCH(TRUE(),
    [Variation Taux Retour] > 0.5,  "#E24B4A",
    [Variation Taux Retour] < -0.5, "#1D9E75",
    "#888780"
)
```

Famille B (rouge si ↑) : `Couleur Variation Taux Retour`, `Couleur Variation Annulation`, `Couleur Variation Frais`, `Couleur Variation Delai`, `Couleur Variation Abandon`, `Couleur Variation CSAT`, `Couleur Variation Avis Negatif`, `Couleur Variation Stock Rupture`, `Couleur Variation Refund Time`.

## 4.6 Dossier `06 — Funnel digital` (10 mesures)

> **🎯 Rôle du dossier** : décrire la conversion étape par étape, du visiteur unique à l'achat.

### 📘 Concept clé — le funnel comme cascade

Imagine 100 visiteurs qui entrent sur la home. Combien voient un produit ? Combien ajoutent au panier ? Combien finalisent ? Le funnel répond à ces 4 questions.

```
  Visiteurs uniques  ------  100 % (référence)
       |
       v
  Page produit vue   ------  ~62 %
       |
       v
  Ajout au panier    ------  ~28 %
       |
       v
  Achat complété     ------  ~7 %
```

### Les 10 mesures

```dax
Visiteurs Uniques  = DISTINCTCOUNT(fact_web_logs[session_id])
Sessions Totales   = COUNTROWS(fact_web_logs)
Pages Vues Total   = SUM(fact_web_logs[page_views])
Pages Vues Moyenne = AVERAGE(fact_web_logs[page_views])
Duree Session Moy  = AVERAGE(fact_web_logs[session_duration])
Sessions Produit   = CALCULATE([Sessions Totales], fact_web_logs[viewed_product] = TRUE)
Sessions Panier    = CALCULATE([Sessions Totales], fact_web_logs[added_to_cart] = TRUE)
Sessions Achat     = CALCULATE([Sessions Totales], fact_web_logs[purchased] = TRUE)
Taux Conversion    = DIVIDE([Sessions Achat], [Visiteurs Uniques])
Taux Abandon Panier= DIVIDE([Sessions Panier] - [Sessions Achat], [Sessions Panier])
```

### ⚠️ Piège fréquent — le grain du funnel

Le funnel se calcule **par session**, pas par utilisateur. Un même client revient plusieurs fois et chaque visite est un événement indépendant. Si tu confonds, tes taux deviennent supérieurs à 100 % et tu perds toute crédibilité en réunion.

# V — Créer les colonnes calculées

> **🎯 Objectif** : ajouter deux colonnes calculées qui résolvent des cas réels que les mesures seules ne peuvent pas traiter.

### 📘 Concept clé — colonne calculée vs mesure (la différence qui sauve la vie)

| | Colonne calculée | Mesure |
|---|---|---|
| **Quand est-elle évaluée ?** | Au refresh, ligne par ligne | À chaque rendu de visuel |
| **Où est-elle stockée ?** | Dans la table | Nulle part (recalculée à la volée) |
| **Peut-elle servir de filtre / légende ?** | Oui | Non (sauf rares exceptions) |
| **Coût mémoire** | + (taille du modèle augmente) | 0 |
| **Coût CPU au visuel** | 0 | + (recalcul à chaque slicer) |

**Règle de pouce.** Si tu as besoin de la valeur comme **slicer, légende, axe ou catégorie**, c'est une **colonne calculée**. Sinon, c'est une **mesure**.

C'est exactement la situation de `Categorie_Groupee` (utilisée comme légende d'un graphique en barres) : elle ne pouvait pas être une mesure.

## 5.1 `dim_products[Categorie_Groupee]` — top 4 + Autres

> **🎯 À la fin de cette section**, ton graphique "Répartition par catégorie" affichera les 4 plus grosses catégories nominales et regroupera toutes les autres dans une barre "Autres".

### 📘 Le problème métier

ShopAfrica+ a 12 catégories de produits. Un graphique avec 12 barres devient illisible. Le client veut voir les **4 catégories qui font 80 % du CA** et regrouper le reste.

### 📘 Pseudocode → DAX

Avant d'écrire le DAX, écris ce que tu veux faire en français :

```
POUR chaque produit :
    SOIT _ca = CA de la catégorie de ce produit
    SOIT _allCats = liste de toutes les catégories avec leur CA
    SOIT _rang = rang de _ca dans _allCats (1 = meilleure)
    SI _rang <= 4 ALORS catégorie de ce produit
    SINON "Autres"
```

### Le DAX

**Vue Données → table `dim_products` → Nouvelle colonne** :

```dax
Categorie_Groupee = 
VAR _ca_current = 
    CALCULATE(
        SUM(fact_order_items[line_revenue]),
        fact_orders[order_status] = "Livree",
        ALLEXCEPT(dim_products, dim_products[categorie])
    )
VAR _allCats = 
    ADDCOLUMNS(
        ALL(dim_products[categorie]),
        "@ca",
        CALCULATE(
            SUM(fact_order_items[line_revenue]),
            fact_orders[order_status] = "Livree",
            ALLEXCEPT(dim_products, dim_products[categorie])
        )
    )
VAR _rank = COUNTROWS(FILTER(_allCats, [@ca] > _ca_current)) + 1
RETURN
IF(_rank <= 4, dim_products[categorie], "Autres")
```

### ⚠️ Le piège que 9 personnes sur 10 manquent

Sans `ALLEXCEPT(dim_products, dim_products[categorie])` dans le `CALCULATE` interne, le contexte de ligne propage le `product_id` courant et tu ne calcules le CA que de **ce** produit, pas de toute sa catégorie. Résultat : tous les produits sont rangés "Autres" car aucun produit individuel ne dépasse les 4 premiers.

**Mémoriser cette règle** : quand une colonne calculée appelle un agrégat *au niveau d'un sous-groupe*, mettre un `ALLEXCEPT` qui garde uniquement la dimension de regroupement.

## 5.2 `fact_web_logs[session_date]` — relier les sessions au Calendrier

> **🎯 À la fin de cette section**, le slicer Année / Mois sur la page Funnel filtrera correctement les sessions web.

### 📘 Le problème

La colonne `session_timestamp` est un `DateTime` à la milliseconde. Or, pour relier au Calendrier, il faut une `Date` pure (sans heure). Sans cette colonne, la relation Calendrier ↔ fact_web_logs est impossible et tous les filtres temporels échouent.

### Le DAX

**Vue Données → table `fact_web_logs` → Nouvelle colonne** :

```dax
session_date = 
DATE(
    YEAR(fact_web_logs[session_timestamp]),
    MONTH(fact_web_logs[session_timestamp]),
    DAY(fact_web_logs[session_timestamp])
)
```

Puis crée la relation `Calendrier[Date]` (1) → `fact_web_logs[session_date]` (N).

### 🚀 Pour aller plus loin

Pourquoi ne pas faire `DATEVALUE(session_timestamp)` qui ferait pareil en plus court ? Parce que `DATEVALUE` rate certains formats locaux (FR vs US) et plante silencieusement. Le pattern `DATE(YEAR, MONTH, DAY)` est universel.

## 9.2 Pièges fréquents et leurs solutions

| Symptôme | Cause probable | Correction |
|---|---|---|
| `SAMEPERIODLASTYEAR` renvoie blank | `Calendrier` non marquée comme table de dates | Onglet *Outils de table* → Marquer comme table de dates |
| Le total dans une carte est faux | Filtre actif oublié sur le visuel ou le slicer | Vérifier le volet *Filtres* du visuel |
| Le scatter refuse une mesure en légende | Une mesure ne peut pas être une légende | Convertir en **colonne calculée** |
| `SELECTEDVALUE` renvoie blank dans un total | Plusieurs valeurs dans le contexte | Remplacer par `MAX` ou `MIN` selon le sens |
| `DIVIDE` renvoie blank au lieu de 0 | Voulu par défaut, mais peut être indésirable | `DIVIDE(num, den, 0)` pour forcer le 0 |
| Variation reste à 0 même quand la valeur change | Comparer une mesure à elle-même | Vérifier la mesure N-1 et son `SAMEPERIODLASTYEAR` |
| Le visuel ne se rafraîchit pas | Cache figé | **Accueil → Actualiser** ou rouvrir le `.pbix` |
| Les emojis ▲ ▼ sont mal rendus | Police par défaut sans support Unicode | Forcer la police **Segoe UI Symbol** sur le champ texte |

## 9.3 Storytelling exécutif

Quand tu présenteras ce dashboard à un comité de direction, suis cet ordre narratif :

1. **Constat global** (page 1) : « Le CA progresse de +18 % vs N-1, mais les commandes seulement de +9 % — donc le panier moyen explose. »
2. **Causes côté produit** (page 2) : « 4 catégories portent 80 % du CA, et 2 produits stars représentent 15 % à eux seuls. »
3. **Causes côté client** (page 3) : « Les VIP achètent 3× plus que les Réguliers ; les A-risque (12 % du parc) ont un taux de retour 3× plus élevé. »
4. **Causes côté funnel** (page 4) : « 28 % des visiteurs ajoutent au panier, mais 75 % d'entre eux abandonnent — l'abandon panier est notre plus gros levier. »
5. **Causes côté logistique** (page 5) : « San-Pédro a un délai moyen de 6,8 j vs 3,2 j à Abidjan — explication probable de l'écart de satisfaction. »
6. **Recommandations** (3 actions chiffrées) : projection à 6 mois si on agit.

## 9.4 Annexes

### Règles DAX universelles à graver

1. Toujours utiliser `DIVIDE` pour diviser
2. Pattern `VAR ... RETURN` même pour les mesures simples
3. Utiliser `SELECTEDVALUE` ou `MAX` au lieu de `VALUES` quand on attend une valeur unique
4. Dans `CALCULATE`, écrire `Table[Col] = "value"` plutôt que `Mesure = "value"`
5. Pour les variations en pp (points de pourcentage), faire une soustraction, pas un `DIVIDE`
6. `ALLEXCEPT` quand une colonne calculée appelle un agrégat à un niveau de dimension
