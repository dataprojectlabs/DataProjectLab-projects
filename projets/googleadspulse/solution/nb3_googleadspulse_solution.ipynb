et Barbara Minto pour structurer une narrative persuasive en 5 slides :

```
Slide 1 : Situation      — Où en sommes-nous aujourd'hui ?
Slide 2 : Complication   — Quel est le problème concret ?
Slide 3 : Question       — Quelle décision doit-on prendre ?
Slide 4 : Réponse        — Quelle est la recommandation chiffrée ?
Slide 5 : Action         — Quel est le plan d'exécution ?
```

## 2. Architecture Power BI en étoile

### 🎓 MÉTHODE — Le modèle en étoile

En Power BI, l'architecture **étoile** (star schema) est **la seule bonne approche** pour
les dashboards performants. Principe :
- **1 table de faits** au centre (les métriques numériques, 1 ligne = 1 événement mesurable)
- **N tables de dimensions** autour (les contextes : temps, campagne, compte, device...)
- **Relations 1:N** uniquement depuis les dimensions vers les faits

**Pourquoi ça compte :**
1. **Performance** — les filtres se propagent en cascade, les visuels se chargent en < 1 seconde
2. **Mesures DAX simples** — les relations font le travail, pas besoin de jointures manuelles
3. **Maintenance** — ajouter une dimension ne casse rien

### Architecture de GoogleAdsPulse

```
                      ┌─────────────────────┐
                      │   dim_calendrier    │ ◄─── colonne calculée CALENDAR()
                      │  date · mois · an   │
                      │  semaine · trimestre │
                      └──────────┬──────────┘
                                 │ 1:N
┌──────────────┐    ┌─────────────▼────────────┐    ┌──────────────┐
│ dim_accounts │───►│                          │◄───│ dim_campaigns│
│   5 lignes   │    │    fact_performance      │    │   25 lignes  │
└──────────────┘    │       (41 428 lignes)    │    └──────────────┘
                    │                          │
┌──────────────┐    │  date · account · camp   │    ┌──────────────┐
│ dim_keywords │───►│  ad · keyword · device   │◄───│   dim_ads    │
│  372 lignes  │    │  country · hour · dow    │    │  121 lignes  │
└──────────────┘    │                          │    └──────────────┘
                    │  impressions · clicks    │
                    │  cost · conversions      │
                    │  conversion_value        │
                    └──────────────────────────┘
```

Les 7 CSV analytiques du NB2 sont importés comme **tables auxiliaires** (pas dans l'étoile,
mais consommées directement par les visuels). Exemple : `gads_cohort_keywords.csv` alimente
uniquement la heatmap de la page Keywords.


### Étapes d'import dans Power BI Desktop

**1. Connexion aux sources**
```
Accueil → Obtenir les données → Texte/CSV
  - accounts.csv
  - campaigns.csv
  - ads.csv
  - keywords.csv
  - performance_quotidienne.csv  (la table de faits)
  
  Puis les 7 CSV analytiques NB2 :
  - gads_kpi_mensuel.csv
  - gads_campaigns_rank.csv
  - gads_rolling.csv
  - gads_cohort_keywords.csv
  - gads_anomalies.csv
  - gads_jour_heure.csv
  - gads_benchmark.csv
```

**2. Nettoyage minimal dans Power Query**
- `performance_quotidienne` : filtrer `cost_eur >= 0` et `clicks <= impressions`
- Convertir `date` en type Date (pas DateTime)
- Supprimer les doublons sur la clé composite (date, campaign_id, ad_id, keyword_id, device, hour)

**3. Créer la table calendrier (mesure DAX)**
```DAX
dim_calendrier = 
ADDCOLUMNS(
    CALENDAR(DATE(2023,1,1), DATE(2024,12,31)),
    "Annee",      YEAR([Date]),
    "Mois_Num",   MONTH([Date]),
    "Mois_Nom",   FORMAT([Date], "MMMM"),
    "Trimestre",  "T" & FORMAT([Date], "Q"),
    "Semaine",    WEEKNUM([Date], 2),
    "Jour_Num",   WEEKDAY([Date], 2),
    "Jour_Nom",   FORMAT([Date], "dddd"),
    "AnneeMois",  FORMAT([Date], "yyyy-MM")
)
```

**4. Créer les relations**
```
dim_calendrier[Date]      ──1:N──► fact_performance[date]
dim_accounts[account_id]  ──1:N──► fact_performance[account_id]
dim_campaigns[campaign_id]──1:N──► fact_performance[campaign_id]
dim_ads[ad_id]            ──1:N──► fact_performance[ad_id]
dim_keywords[keyword_id]  ──1:N──► fact_performance[keyword_id]
```

**5. Marquer la table calendrier comme table de dates**
Clic droit sur `dim_calendrier` → *Marquer comme table de dates* → sélectionner `Date`.
Indispensable pour que les fonctions DAX temporelles (DATEADD, SAMEPERIODLASTYEAR, etc.)
fonctionnent correctement.

## 3. Mesures DAX fondamentales (10 mesures)

### 🎓 MÉTHODE — Mesures vs colonnes calculées

En DAX, on distingue :
- **Colonnes calculées** : calculées ligne par ligne au moment de l'import (lourdes)
- **Mesures** : calculées dynamiquement selon le contexte de filtre des visuels (légères)

**Règle d'or :** pour les KPIs agrégés (sommes, moyennes, ratios), toujours utiliser des
**mesures**. Jamais de colonnes calculées.

**Syntaxe minimale :**
```DAX
NomDeLaMesure = EXPRESSION_DAX
```

### Groupe 1 — Mesures de volume (3 mesures)

```DAX
-- Mesure 1
Total Impressions = SUM(fact_performance[impressions])

-- Mesure 2
Total Clicks = SUM(fact_performance[clicks])

-- Mesure 3
Total Conversions = SUM(fact_performance[conversions])
```

### Groupe 2 — Mesures d'efficacité (5 mesures)

```DAX
-- Mesure 4 : CTR — toujours protéger contre division par zéro
CTR = 
DIVIDE(
    [Total Clicks],
    [Total Impressions],
    0
)

-- Mesure 5 : CPC
CPC = 
DIVIDE(
    SUM(fact_performance[cost_eur]),
    [Total Clicks],
    0
)

-- Mesure 6 : CPM (cost per 1000 impressions)
CPM = 
DIVIDE(
    SUM(fact_performance[cost_eur]) * 1000,
    [Total Impressions],
    0
)

-- Mesure 7 : Cost per Conversion
Cost per Conversion = 
DIVIDE(
    SUM(fact_performance[cost_eur]),
    [Total Conversions],
    0
)

-- Mesure 8 : Conversion Rate
Conversion Rate = 
DIVIDE(
    [Total Conversions],
    [Total Clicks],
    0
)
```

### Groupe 3 — Mesures de rentabilité (2 mesures + ROAS)

```DAX
-- Mesure 9 : Spend
Total Spend = SUM(fact_performance[cost_eur])

-- Mesure 10 : Conversions Value
Conversions Value = SUM(fact_performance[conversion_value_eur])

-- Mesure bonus : ROAS
ROAS = 
DIVIDE(
    [Conversions Value],
    [Total Spend],
    0
)
```

**🏢 ACTION MÉTIER :** ces 10 mesures sont la base de **toutes les pages du dashboard**.
Elles se combinent automatiquement avec n'importe quel filtre (période, campagne, device)
sans avoir à réécrire la logique.

## 4. Mesures DAX avancées — les deltas period-over-period (20+ mesures)

### 🎓 MÉTHODE — Comparer à la période précédente

Le *«+3.41%»* / *«-11.42%»* à côté des KPIs cards est calculé avec la fonction `DATEADD()`
qui décale le contexte de filtre temporel :

```DAX
-- Pattern général : valeur de la période PRÉCÉDENTE
Mesure Prev = 
CALCULATE(
    [Mesure Actuelle],
    DATEADD(dim_calendrier[Date], -1, MONTH)
)

-- Pattern delta % : variation vs période précédente
Mesure Delta % = 
DIVIDE(
    [Mesure Actuelle] - [Mesure Prev],
    [Mesure Prev],
    0
)
```

### 4.1 — Valeurs période précédente (10 mesures)

```DAX
Impressions Prev = 
CALCULATE([Total Impressions], DATEADD(dim_calendrier[Date], -1, MONTH))

Clicks Prev = 
CALCULATE([Total Clicks], DATEADD(dim_calendrier[Date], -1, MONTH))

Conversions Prev = 
CALCULATE([Total Conversions], DATEADD(dim_calendrier[Date], -1, MONTH))

CTR Prev = 
CALCULATE([CTR], DATEADD(dim_calendrier[Date], -1, MONTH))

CPC Prev = 
CALCULATE([CPC], DATEADD(dim_calendrier[Date], -1, MONTH))

CPM Prev = 
CALCULATE([CPM], DATEADD(dim_calendrier[Date], -1, MONTH))

Cost per Conversion Prev = 
CALCULATE([Cost per Conversion], DATEADD(dim_calendrier[Date], -1, MONTH))

Conversion Rate Prev = 
CALCULATE([Conversion Rate], DATEADD(dim_calendrier[Date], -1, MONTH))

Total Spend Prev = 
CALCULATE([Total Spend], DATEADD(dim_calendrier[Date], -1, MONTH))

Conversions Value Prev = 
CALCULATE([Conversions Value], DATEADD(dim_calendrier[Date], -1, MONTH))
```

### 4.2 — Deltas % (10 mesures)

```DAX
Impressions Delta % = 
DIVIDE([Total Impressions] - [Impressions Prev], [Impressions Prev], 0)

Clicks Delta % = 
DIVIDE([Total Clicks] - [Clicks Prev], [Clicks Prev], 0)

Conversions Delta % = 
DIVIDE([Total Conversions] - [Conversions Prev], [Conversions Prev], 0)

CTR Delta % = 
DIVIDE([CTR] - [CTR Prev], [CTR Prev], 0)

CPC Delta % = 
DIVIDE([CPC] - [CPC Prev], [CPC Prev], 0)

-- etc. pour CPM, CostPerConv, ConvRate, Spend, ConvValue
```

### 4.3 — Formatage conditionnel couleurs

Pour afficher le delta en **vert si positif, rouge si négatif**, on crée une mesure qui
retourne un code hexadécimal couleur :

```DAX
-- Pour KPIs où MONTÉE est bonne (Impressions, Clicks, Conversions, ROAS)
Delta Color Good Up = 
IF([Conversions Delta %] >= 0, "#1D9E75", "#E24B4A")

-- Pour KPIs où MONTÉE est mauvaise (CPC, CPM, Cost/Conv)
Delta Color Good Down = 
IF([CPC Delta %] <= 0, "#1D9E75", "#E24B4A")
```

**🏢 UTILISATION :** dans la visualisation KPI card → *Formatage conditionnel sur la couleur
de la valeur du delta* → Par règle → Format : "Valeur de champ" → Champ : la mesure couleur.

**⚠️ Subtilité métier :** pour le CPC, une baisse est BONNE (on paye moins cher pour un clic).
Pour les Impressions, une hausse est généralement BONNE. La fonction couleur doit s'adapter
au KPI — ne jamais réutiliser la même logique pour tout.


### 4.4 — Mesures contextuelles supplémentaires (10 mesures)

```DAX
-- ROAS seuil : la mesure devient un label interprétable
ROAS Verdict = 
SWITCH(
    TRUE(),
    [ROAS] >= 5, "⭐ Très rentable",
    [ROAS] >= 3, "✅ Rentable",
    [ROAS] >= 1, "⚠️ Limite",
    "❌ Non rentable"
)

-- Moyenne glissante 7 jours (pour les timelines lissées)
Spend 7d MA = 
AVERAGEX(
    DATESINPERIOD(
        dim_calendrier[Date],
        LASTDATE(dim_calendrier[Date]),
        -7,
        DAY
    ),
    [Total Spend]
)

-- Spend cumulé depuis le début d'année
Spend YTD = 
TOTALYTD([Total Spend], dim_calendrier[Date])

-- Nombre de campagnes actives dans le contexte
Nb Campagnes Actives = 
CALCULATE(
    DISTINCTCOUNT(fact_performance[campaign_id]),
    fact_performance[clicks] > 0
)

-- Part du spend sur le total de l'account (utile pour drill-down)
% Spend Account = 
DIVIDE(
    [Total Spend],
    CALCULATE([Total Spend], ALL(dim_campaigns)),
    0
)

-- Anomalie flag : est-ce que la campagne est dans la liste des anomalies ?
Is Anomalie = 
IF(
    ISBLANK(RELATED(gads_anomalies[niveau_alerte])),
    "",
    RELATED(gads_anomalies[niveau_alerte])
)

-- Quality Score moyen des keywords (par campagne)
QS Moyen = 
AVERAGE(dim_keywords[quality_score])

-- Total conversions de type spécifique (exemple : seulement Purchase)
Purchase Conversions = 
CALCULATE(
    [Total Conversions],
    fact_performance[conversion_type] = "Purchase"
)

-- Lead conversions
Lead Conversions = 
CALCULATE(
    [Total Conversions],
    fact_performance[conversion_type] IN {"Lead", "Form Submit", "Phone Call"}
)

-- Year-over-Year pour comparer 2024 à 2023
Spend YoY % = 
VAR SpendPrevYear = 
    CALCULATE([Total Spend], SAMEPERIODLASTYEAR(dim_calendrier[Date]))
RETURN
    DIVIDE([Total Spend] - SpendPrevYear, SpendPrevYear, 0)
```

**Récapitulatif :** on a maintenant **30+ mesures** qui couvrent tous les besoins des 5
pages du dashboard, avec deltas automatiques et verdicts interprétables.

### 💡 Points de vigilance design

1. **Couleurs deltas cohérentes** — définir UNE fois la règle *«vert = bon, rouge = mauvais»*

2. **Axes secondaires** — pour les graphiques *dual-axis*, forcer les unités lisibles
   (axe 1 en K, axe 2 en €) pour éviter les écrasements de courbes.

3. **Échelles partagées** — les 4 timelines doivent avoir le **même axe X** (la période
   sélectionnée dans le filtre). Power BI le fait automatiquement si la dimension temporelle
   est la table `dim_calendrier`.

4. **Format des nombres** — 82 516 impressions s'écrit `82,51K` (format condensé), 12 080 €
   s'écrit `12,08K €`. Dans Power BI : *Format > Personnalisé > Unités > Milliers*.

```
┌─────────────────────────────────────────────────────────────────────┐
│ Page 4 — Conversions                                                 │
├─────────────────────────────────────────────────────────────────────┤
│ ┌───────────────────────────┐  ┌───────────────────────────┐         │
│ │ CONVERSIONS PAR TYPE      │  │ COST PER CONVERSION / TYPE│         │
│ │  (donut chart)            │  │  (bar chart horizontal)   │         │
│ │                           │  │                           │         │
│ │  Purchase  40%            │  │  Purchase     12€         │         │
│ │  Lead      25%            │  │  Lead         18€         │         │
│ │  Signup    15%            │  │  Signup        8€         │         │
│ │  Demo Req. 10%            │  │  Demo Req.    25€         │         │
│ │  Form Sub.  6%            │  │  Form Sub.    15€         │         │
│ │  Phone Call 4%            │  │  Phone Call    9€         │         │
│ └───────────────────────────┘  └───────────────────────────┘         │
│                                                                       │
│ ┌─────────────────────────────────────────────────────────────────┐ │
│ │ ÉVOLUTION CONVERSIONS × TYPE (24 mois)                          │ │
│ │ (stacked area chart)                                             │ │
│ └─────────────────────────────────────────────────────────────────┘ │
│                                                                       │
│ ┌───────────────────────────┐  ┌───────────────────────────┐         │
│ │ CONV VALUE / TYPE         │  │ CONVERSIONS × CAMPAIGN    │         │
│ │  (bar chart)              │  │  (tableau)                │         │
│ │                           │  │                           │         │
│ │  Purchase   2.1M€         │  │  Top 10 campagnes         │         │
│ │  Lead       850K€         │  │  par nb de conv           │         │
│ │  ...                      │  │                           │         │
│ └───────────────────────────┘  └───────────────────────────┘         │
└─────────────────────────────────────────────────────────────────────┘
```

### Mesures dédiées

```DAX
-- Déjà créées en partie 4.4 :
-- Purchase Conversions
-- Lead Conversions

-- Nouvelles mesures pour cette page :
Signup Conversions = 
CALCULATE([Total Conversions], fact_performance[conversion_type] = "Signup")

Demo Conversions = 
CALCULATE([Total Conversions], fact_performance[conversion_type] = "Demo Request")

Purchase Value = 
CALCULATE([Conversions Value], fact_performance[conversion_type] = "Purchase")

-- Cost per conversion type (avec allocation proportionnelle)
Cost per Purchase = 
DIVIDE(
    [Total Spend] * DIVIDE([Purchase Conversions], [Total Conversions], 0),
    [Purchase Conversions],
    0
)
```

### 🏢 ACTION MÉTIER

Pour les clients **e-commerce** (TechShop, AfriHotels), Purchase domine et sa conversion
value directe valide le ROAS. Pour les clients **B2B** (BankAfrica, MedSupply), Lead +
Demo Request dominent et la conversion value est *estimée* (panier moyen × taux de transfo
du lead en client).
